```python
import pickle
import numpy as np

# data/waveforms.pickle ファイルを開く
with open("../../data/waveforms.pickle", "rb") as file:
    waveforms = pickle.load(file)

# データの基本情報を確認
print("=== Waveforms データの詳細 ===")
print(f"データ型: {type(waveforms)}")
print(f"リストの長さ: {len(waveforms)}")

# 最初の要素を確認
if len(waveforms) > 0:
    first_item = waveforms[0]
    print(f"\n最初の要素の型: {type(first_item)}")

    # NumPy配列の場合
    if isinstance(first_item, np.ndarray):
        print(f"最初の要素の形状: {first_item.shape}")
        print(f"最初の要素のデータ型: {first_item.dtype}")

        # すべての要素が同じ形状か確認
        shapes = [item.shape for item in waveforms[: min(5, len(waveforms))]]
        print(f"\n最初の5個の要素の形状: {shapes}")

        # 統計情報（最初の要素）
        print("\n=== 最初の要素の統計情報 ===")
        print(f"最小値: {first_item.min():.6f}")
        print(f"最大値: {first_item.max():.6f}")
        print(f"平均値: {first_item.mean():.6f}")
        print(f"標準偏差: {first_item.std():.6f}")

        # 最初の数値を表示
        print(f"\n最初の10個の値: {first_item.flatten()[:10]}")
    else:
        # その他のデータ型の場合
        print(f"最初の要素の内容（最初の100文字）: {str(first_item)[:100]}")

        # リストやタプルの場合
        if isinstance(first_item, (list, tuple)):
            print(f"最初の要素の長さ: {len(first_item)}")
            if len(first_item) > 0:
                print(f"最初の要素の最初の項目: {first_item[0]}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# 日本語フォント
font_path = "/System/Library/Fonts/ヒラギノ角ゴシック W3.ttc"
font_prop = fm.FontProperties(fname=font_path)
plt.rcParams["font.family"] = font_prop.get_name()
plt.rcParams["axes.unicode_minus"] = False
```

```python
import matplotlib.pyplot as plt

def plot_waveform(waveform_idx, waveforms):
    """指定されたインデックスの波形をプロットする関数"""
    # waveforms[0]とwaveforms[1]の指定された波形を取得
    waveform_0 = waveforms[0][waveform_idx]  # shape: (2250, 2)
    waveform_1 = waveforms[1][waveform_idx]  # shape: (251, 1)

    # データのラベル
    data_labels = ["vp", "vs", "iL"]

    # 図を作成
    fig, axes = plt.subplots(2, 2, figsize=(14, 8))
    fig.suptitle(f"波形データ ({waveform_idx}番目の波形) - ドット表示", fontsize=14)

    # waveforms[0]の波形を描画（2チャンネル）
    for ch in range(2):
        ax = axes[0, ch]
        ax.scatter(range(len(waveform_0[:, ch])), waveform_0[:, ch], s=1, color="blue", alpha=0.7)
        ax.set_title(
            f"{data_labels[ch]} (waveforms[0]) - {waveform_idx}番目の波形 - チャンネル {ch + 1}",
            fontsize=12,
        )
        ax.set_xlabel("サンプル")
        ax.set_ylabel("振幅")
        ax.grid(True, alpha=0.3)
        ax.set_ylim(-250, 250)

    # waveforms[1]の波形を描画（1チャンネル）
    ax = axes[1, 0]
    ax.scatter(range(len(waveform_1[:, 0])), waveform_1[:, 0], s=1, color="green", alpha=0.7)
    ax.set_title(
        f"{data_labels[2]} (waveforms[1]) - {waveform_idx}番目の波形 - チャンネル 1", fontsize=12
    )
    ax.set_xlabel("サンプル")
    ax.set_ylabel("振幅")
    ax.grid(True, alpha=0.3)

    # 右下のサブプロットは空にする
    axes[1, 1].axis("off")

    plt.tight_layout()
    plt.show()

    print(f"{waveform_idx}番目の波形描画完了:")
    print(f"- {data_labels[0]} (waveforms[0][{waveform_idx}])の形状: {waveform_0.shape}")
    print(f"- {data_labels[1]} (waveforms[1][{waveform_idx}])の形状: {waveform_1.shape}")

# 12種類すべての波形をプロット
print("=== 12種類すべての波形をプロット開始 ===")
for i in range(12):
    print(f"\n--- {i}番目の波形をプロット中 ---")
    plot_waveform(i, waveforms)

print("\n=== すべての波形のプロット完了 ===")
```
